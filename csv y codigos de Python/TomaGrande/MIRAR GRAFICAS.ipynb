```python
import csv
import matplotlib.pyplot as plt

# Función para plotear un archivo CSV
def plotear_archivo(archivo):
    with open(archivo, 'r') as f:
        next(f)  # Saltar la primera fila (encabezado)
        reader = csv.reader(f)
        datos = [float(row[0]) for row in reader]

    plt.figure(figsize=(10, 6))
    plt.plot(datos)
    plt.xlabel('Muestras')
    plt.ylabel('Datos')
    plt.title(archivo)
    plt.show()

# Lista de números de archivo que quieres plotear (del 1 al 20, por ejemplo)
numeros_archivos = range(100, 200)

# Plotear los archivos seleccionados
for num in numeros_archivos:
    archivo = f"datos6s_EMG115200TOMAindice{num}.csv"
    plotear_archivo(archivo)
```
