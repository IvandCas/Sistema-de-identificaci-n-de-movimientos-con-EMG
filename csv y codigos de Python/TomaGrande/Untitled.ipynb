```python
import pandas as pd
import numpy as np
from math import sqrt

MAV_total=[]
VAR_total=[]
RMS_total=[]
Dedo_indice= []

for i in range(1, 200):
    df = pd.read_csv(f"datos6s_EMG115200TOMAindice{i}.csv")

    df["Dato"]=df["Dato"].astype(int)

    x_n=sum(abs(df["Dato"]))
    print(f'Resultado de sumatoria: {x_n}')
    N=len(df["Dato"])
    print(f'Numero de muestras: {N}')
    MAV=x_n/N
    print(f'Valor medio absoluto: {MAV}')
    MAV_total.append(MAV)

    x_n=sum(abs(df["Dato"]**2))
    print(f'Resultado de sumatoria: {x_n}')
    N=len(df["Dato"])
    print(f'Numero de muestras: {N}')
    VAR=x_n/(N-1)
    print(f'Varianza: {VAR}')
    VAR_total.append(VAR)

    x_n=sum(df["Dato"]**2)
    print(f'Resultado de sumatoria: {x_n}')
    N=len(df["Dato"])
    print(f'Numero de muestras: {N}')
    RMS=sqrt(x_n/N)
    print(f'RMS: {RMS}')
    RMS_total.append(RMS)

    Dedo_indice.append("Dedo indice")

# Crear un DataFrame con los resultados
results_df = pd.DataFrame({
    'MAV_total': MAV_total,
    'VAR_total': VAR_total,
    'RMS_total': RMS_total,
    'Dedo_indice': Dedo_indice
})

# Guardar el DataFrame en un archivo CSV
results_df.to_csv('DATASETTOMAINDICE.csv', index=False)
```
