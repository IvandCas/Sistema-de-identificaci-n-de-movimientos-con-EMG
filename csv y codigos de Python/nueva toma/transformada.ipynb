```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from math import sqrt

# Definimos los parámetros de la señal
MAV_total=[]
VAR_total=[]
RMS_total=[]
Wav_Leng = []
Dasdv= []
Damv = []
iav= []
dt=1
Dedo_indice= []

for i in range(1, 2000):
    # Leer los datos del archivo CSV
    df = pd.read_csv(f"datos6s_EMG115200TOMAINDICE{i}.csv")

    df["Dato"]=df["Dato"].astype(int)

    x_n=sum(abs(df["Dato"]))
    N=len(df["Dato"])
    MAV=x_n/N
    MAV_total.append(MAV)

    x_n=sum(abs(df["Dato"]**2))
    N=len(df["Dato"])
    VAR=x_n/(N-1)
    VAR_total.append(VAR)

    x_n=sum(df["Dato"]**2)
    N=len(df["Dato"])
    RMS=sqrt(x_n/N)
    RMS_total.append(RMS)
    
    wl = np.sum(np.abs(np.diff(df["Dato"])))
    Wav_Leng.append(wl)

    diff_signal = np.diff(df["Dato"])  # Diferencias entre valores consecutivos
    dasdv_value = np.sqrt(np.mean(diff_signal**2))
    Dasdv.append(dasdv_value)

    mean_value = np.mean(df["Dato"])  # Calcular el valor medio de la señal
    damv_value = np.mean(np.abs(df["Dato"] - mean_value))
    Damv.append(damv_value)

    iav_value = np.sum(np.abs(df["Dato"])) * dt
    iav.append(iav_value)

    Dedo_indice.append("Dedo indice")

results_df = pd.DataFrame({
    'MAV_total': MAV_total,
    'VAR_total': VAR_total,
    'RMS_total': RMS_total,
    'Wav_Leng':Wav_Leng,
    'Dasdv':Dasdv,
    'Damv':Damv,
    'iav':iav,
    'Dedo_indice': Dedo_indice
})

results_df.to_csv('DATASETTOMAINDICA.csv', index=False)
```
