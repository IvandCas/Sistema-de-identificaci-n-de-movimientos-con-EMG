```python
import serial
import csv
import time
import keyboard

# Función para obtener el nombre del archivo CSV
def obtener_nombre_archivo(indice):
    return f'datos6s_EMG115200TOMAINDICE{indice}.csv'

# Establecer conexión serial con Arduino
arduino = serial.Serial('COM6', 115200)  # Cambia 'COM6' por el puerto al que está conectado tu Arduino

# Contador para el nombre del archivo
indice_archivo = 300

while True:
    # Nombre del archivo CSV
    nombre_archivo = obtener_nombre_archivo(indice_archivo)

    # Abrir el archivo CSV en modo escritura
    with open(nombre_archivo, mode='w', newline='') as archivo_csv:
        # Crear un objeto escritor CSV
        escritor_csv = csv.writer(archivo_csv)
        
        # Escribir encabezados
        escritor_csv.writerow(['Dato'])

        print(f"Iniciando toma de datos en '{nombre_archivo}'...")
        tiempo_inicial = time.time()  # Guardar el tiempo inicial

        while (time.time() - tiempo_inicial) < 4:
            # Leer línea de datos desde Arduino
            datos = arduino.readline().decode().strip()
            tiempo_actual = time.time() - tiempo_inicial
            
            # Guardar los datos en el archivo CSV
            escritor_csv.writerow([datos])

            # Verificar si se presiona la tecla "Escape" para detener la toma de datos
            if keyboard.is_pressed('esc'):
                break

        print(f"Fin de la toma de datos en '{nombre_archivo}'")

    # Incrementar el contador del nombre del archivo
    indice_archivo += 1

    # Verificar si se presionó la tecla "Escape" para detener la toma de datos
    if keyboard.is_pressed('esc'):
        break

# Cerrar la conexión serial
arduino.close()
```

```python
import csv
import random

# Función para obtener el nombre del archivo CSV
def obtener_nombre_archivo(indice):
    return f'datos6s_EMG115200TOMAABIERTA{indice}.csv'

# Generar datos con una media de 310 y una desviación estándar pequeña (por ejemplo, 3)
def generar_datos(n, media=310, desviacion=0.1):
    return [random.gauss(media, desviacion) for _ in range(n)]

# Contador para el nombre del archivo
indice_archivo = 1

# Generar 100 archivos
while indice_archivo <= 100:
    # Nombre del archivo CSV
    nombre_archivo = obtener_nombre_archivo(indice_archivo)

    # Generar 400 datos
    datos = generar_datos(400)

    # Abrir el archivo CSV en modo escritura
    with open(nombre_archivo, mode='w', newline='') as archivo_csv:
        # Crear un objeto escritor CSV
        escritor_csv = csv.writer(archivo_csv)
        
        # Escribir encabezados
        escritor_csv.writerow(['Dato'])

        print(f"Generando datos en '{nombre_archivo}'...")

        # Guardar los datos en el archivo CSV
        for dato in datos:
            escritor_csv.writerow([dato])

    print(f"Archivo '{nombre_archivo}' generado con éxito")

    # Incrementar el contador del nombre del archivo
    indice_archivo += 1
```

```python
import csv
import os

# Función para dividir los datos y guardar en archivos separados
def dividir_y_guardar(datos, nombre_archivo, indice_archivo):
    # Dividir los datos en grupos de 100
    grupos = [datos[i:i+100] for i in range(0, len(datos), 100)]

    # Carpeta para los archivos divididos
    carpeta = "archivos_divididos"
    if not os.path.exists(carpeta):
        os.makedirs(carpeta)

    # Guardar cada grupo como un archivo separado
    for i, grupo in enumerate(grupos[:4]):  # Solo se generarán cuatro archivos
        nuevo_nombre = f'datos6s_EMG115200TOMAANULAR{indice_archivo * 4 + i}.csv'  # Generar nombres secuenciales
        ruta_archivo = os.path.join(carpeta, nuevo_nombre)
        with open(ruta_archivo, mode='w', newline='') as archivo_csv:
            escritor_csv = csv.writer(archivo_csv)
            escritor_csv.writerow(['Dato'])
            escritor_csv.writerows(grupo)

# Iterar sobre los archivos existentes
for indice_archivo in range(1, 201):  # Suponiendo que los archivos van desde 1 a 200
    nombre_archivo = f'datos6s_EMG115200TOMAANULAR{indice_archivo}.csv'  # Nombre del archivo actual

    # Leer los datos del archivo
    with open(nombre_archivo, mode='r', newline='') as archivo_csv:
        lector_csv = csv.reader(archivo_csv)
        datos = list(lector_csv)[1:]  # Saltar el encabezado y leer los datos

    # Llamar a la función para dividir y guardar los datos
    dividir_y_guardar(datos, nombre_archivo, indice_archivo)

print("Proceso completado.")
```
