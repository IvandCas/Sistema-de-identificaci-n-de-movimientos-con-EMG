```python
import pandas as pd
import numpy as np
from math import sqrt
import serial
import time
from sklearn.tree import DecisionTreeClassifier
import joblib

# Función para calcular MAV, VAR y RMS
def calculate_features(data):
    N = len(data)
    
    MAV = sum(abs(data)) / N
    VAR = sum((data - np.mean(data))**2) / (N - 1)
    RMS = sqrt(sum(data**2) / N)
    
    return MAV, VAR, RMS

# Configurar conexión serial
arduino = serial.Serial('COM6', 115200)
# Cargar el modelo entrenado
#model = joblib.load('modelo_entrenado.pkl')
model = joblib.load('modelo_knn_entrenado.pkl')
#model = joblib.load('modelo_tree_entrenado.pkl')

# Leer datos en bloques de 50
buffer_size = 100
data_buffer = []

try:
    while True:
        if arduino.in_waiting > 0:
            data = arduino.readline().decode('utf-8').strip()
            data_buffer.append(int(data))

            if len(data_buffer) == buffer_size:
                # Calcular características
                MAV, VAR, RMS = calculate_features(np.array(data_buffer))
                
                # Crear DataFrame con las características
                features = pd.DataFrame([[MAV, VAR, RMS]], columns=['MAV_total', 'VAR_total', 'RMS_total'])
                
                # Realizar predicción
                prediction = model.predict(features)
                
                # Imprimir la predicción
                print(f'Predicción: {prediction[0]}')
                
                # Limpiar el buffer
                data_buffer = []

except KeyboardInterrupt:
    print("Interrumpido por el usuario")

finally:
    arduino.close()
```
